Q-Learning is an **off-policy** temporal difference (TD) control algorithm that learns the optimal action-value function Q*(s,a) directly, without requiring a model of the environment.

## The Q-Learning Update Rule

The core of Q-Learning is its update rule:

$$Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$$

Where:
- $s$ = current state
- $a$ = action taken
- $r$ = reward received
- $s'$ = next state
- $\alpha$ = learning rate (how much to update)
- $\gamma$ = discount factor (how much to value future rewards)
- $\max_{a'} Q(s',a')$ = maximum Q-value for the next state

The term in brackets is called the **TD error** (temporal difference error).

## 3 - Epsilon-Greedy Policy

The **exploration-exploitation dilemma** is fundamental in RL:
- **Exploitation**: Use current knowledge to maximize reward
- **Exploration**: Try new actions to discover better strategies

The **ε-greedy policy** balances these:
- With probability $\epsilon$: choose a random action (explore)
- With probability $1-\epsilon$: choose the best known action (exploit)

$$
a = \begin{cases}
\arg\max_{a'} Q(s, a') & \text{with probability } 1-\epsilon \\
\text{random action} & \text{with probability } \epsilon
\end{cases}
$$

Typically, we start with high $\epsilon$ (e.g., 1.0) and decay it over time.

## 4 - Q-Learning Update

Now we implement the core Q-Learning update rule. After taking action $a$ in state $s$ and observing reward $r$ and next state $s'$, we update:

$$Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$$

Let's break this down:
1. **Current estimate**: $Q(s,a)$
2. **TD target**: $r + \gamma \max_{a'} Q(s',a')$ (immediate reward + discounted best future value)
3. **TD error**: $r + \gamma \max_{a'} Q(s',a') - Q(s,a)$ (difference between target and current)
4. **Update**: Move Q(s,a) toward the target by a fraction $\alpha$

**Special case**: If $s'$ is a terminal state (episode ended), there's no future reward, so:
$$Q(s,a) \leftarrow Q(s,a) + \alpha \left[r - Q(s,a)\right]$$

- El episodio termina cuando `done = terminated or truncated`

````python
# GRADED FUNCTION: run_single_episode

def run_single_episode(env, Q, epsilon, alpha, gamma):
    """
    Ejecuta un episodio completo de Q-Learning.
    
    Arguments:
    env -- ambiente de Gymnasium
    Q -- Q-table actual
    epsilon -- probabilidad de exploración
    alpha -- learning rate
    gamma -- discount factor
    
    Returns:
    Q -- Q-table actualizada
    total_reward -- recompensa total del episodio
    """
    # (approx. 10 lines)
    # 1. Resetear ambiente: state = env.reset()[0]
    # 2. Inicializar done = False y total_reward = 0
    # 3. Loop mientras not done:
    #    a. Elegir acción usando epsilon_greedy_action
    #    b. Tomar acción: next_state, reward, terminated, truncated, _ = env.step(action)
    #    c. done = terminated or truncated
    #    d. Actualizar Q usando q_learning_update
    #    e. total_reward += reward
    #    f. state = next_state
    # 4. Retornar Q y total_reward
    
    ### START CODE HERE ###
    state = env.reset()[0]
    done = False
    total_reward = 0
    
    while not done:
        action = epsilon_greedy_action(Q, state, epsilon)
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        Q = q_learning_update(Q, state, action, reward, next_state, done, alpha, gamma)
        total_reward += reward
        state = next_state
    ### END CODE HERE ###
    
    return Q, total_reward
````

### Exercise 4c - train_q_learning_with_decay

Finalmente, implementa la **versión completa con epsilon decay**.

**¿Por qué epsilon decay?**
- Al principio del entrenamiento, queremos **explorar mucho** (epsilon alto)
- Al final del entrenamiento, queremos **explotar** lo aprendido (epsilon bajo)

**Fórmula de decay lineal**:
```python
epsilon = max(epsilon_end, epsilon_start - (epsilon_start - epsilon_end) * episode / n_episodes)
```

Esta fórmula hace que epsilon disminuya linealmente de `epsilon_start` a `epsilon_end`.

````python
# GRADED FUNCTION: train_q_learning_with_decay

def train_q_learning_with_decay(env, n_episodes, alpha=0.1, gamma=0.99, 
                                epsilon_start=1.0, epsilon_end=0.01):
    """
    Entrena un agente Q-Learning con epsilon decay.
    
    Arguments:
    env -- ambiente de Gymnasium
    n_episodes -- número de episodios de entrenamiento
    alpha -- learning rate
    gamma -- discount factor
    epsilon_start -- epsilon inicial (alta exploración)
    epsilon_end -- epsilon final (baja exploración)
    
    Returns:
    Q -- Q-table entrenada
    rewards -- lista de recompensas por episodio
    """
    # (approx. 10 lines)
    # Similar a train_q_learning_simple, pero:
    # - Antes de cada episodio, calcular epsilon actual usando la fórmula de decay
    # - Pasar ese epsilon a run_single_episode
    
    ### START CODE HERE ###
    Q = initialize_q_table(env.observation_space.n, env.action_space.n)
    rewards = []
    
    for episode in range(n_episodes):
        # Epsilon decay
        epsilon = max(epsilon_end, epsilon_start - (epsilon_start - epsilon_end) * episode / n_episodes)
        Q, reward = run_single_episode(env, Q, epsilon, alpha, gamma)
        rewards.append(reward)
    ### END CODE HERE ###
    
    return Q, rewards
````

````python
# Test your implementation
print("Training agent (with epsilon decay)...")
print("This may take 20-30 seconds...\n")

env_train = gym.make('FrozenLake-v1', is_slippery=False)

Q_final, rewards_final = train_q_learning_with_decay(
    env_train,
    n_episodes=2000,
    alpha=0.1,
    gamma=0.99,
    epsilon_start=1.0,
    epsilon_end=0.01
)

print("Training completed!")
print(f"\nFirst 100 episodes: {np.mean(rewards_final[:100])*100:.0f}% success")
print(f"Last 100 episodes: {np.mean(rewards_final[-100:])*100:.0f}% success")
print(f"Improvement: +{(np.mean(rewards_final[-100:]) - np.mean(rewards_final[:100]))*100:.0f}%")

env_train.close()

# Run the test
train_q_learning_with_decay_test(train_q_learning_with_decay)
````

### 6.3 - Evaluate Agent

````python
def evaluate_agent(env, Q, n_episodes=100):
    """
    Evaluate trained agent (no exploration).
    """
    total_rewards = []
    
    for _ in range(n_episodes):
        state, _ = env.reset()
        episode_reward = 0
        done = False
        
        while not done:
            # Always exploit (epsilon=0)
            action = np.argmax(Q[state])
            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated
            episode_reward += reward
            state = next_state
        
        total_rewards.append(episode_reward)
    
    return np.mean(total_rewards), np.std(total_rewards)

# Evaluate
mean_reward, std_reward = evaluate_agent(env, Q, n_episodes=100)
print(f"\nEvaluation over 100 episodes:")
print(f"Average reward: {mean_reward:.4f} ± {std_reward:.4f}")
print(f"Success rate: {mean_reward * 100:.1f}%")

env.close()
````
