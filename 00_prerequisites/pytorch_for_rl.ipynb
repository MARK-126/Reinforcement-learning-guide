## ¿Qué es PyTorch?

PyTorch es una biblioteca para crear y entrenar redes neuronales. En Deep RL, usamos redes neuronales para:
- Aproximar Q-values (DQN)
- Aproximar políticas (Policy Gradient)
- Aproximar funciones de valor (Actor-Critic)
