En RL, trabajarás constantemente con:

**Ejemplo**: Imagina un juego con 4 estados y 2 acciones (izquierda/derecha).

```

         Izquierda  Derecha
Estado 0:   0.5      0.3
Estado 1:   0.2      0.8
Estado 2:   0.1      0.4
Estado 3:   0.9      0.6
```

En NumPy, esto es un array 2D de forma `(4, 2)`.

Al principio, no sabemos los valores, así que inicializamos todo en **ceros**.

## 3 - Acceder a Valores

**Sintaxis**: `Q[estado, accion]`

## 4 - Seleccionar Mejor Acción

En RL, a menudo queremos seleccionar la **mejor acción** = la acción con el **Q-value más alto**.

Para esto usamos `np.argmax()`.

**Ejemplo**:
```python
valores = np.array([0.3, 0.8, 0.2])
mejor_indice = np.argmax(valores)  # Resultado: 1 (porque 0.8 es el máximo)
```

Para una Q-table, queremos el argmax de una **fila** (un estado):
```python
Q = np.array([[0.5, 0.3],
              [0.2, 0.8]])

mejor_accion = np.argmax(Q[0])  # Fila 0: [0.5, 0.3] → argmax = 0
```

## 5 - Actualizar Q-Values

```python

```

