## ¿Qué es Q-Learning?

Q-Learning es un algoritmo que aprende **qué acción tomar en cada estado** para maximizar recompensas.

### La Idea Central

Imagina que tienes una **tabla de valores** (Q-table) que te dice qué tan buena es cada acción en cada estado:

```
         Acción 0  Acción 1
Estado 0:   0.5      0.8    ← En estado 0, acción 1 es mejor
Estado 1:   0.9      0.2    ← En estado 1, acción 0 es mejor
```

Q-Learning **actualiza esta tabla** cada vez que el agente toma una acción y observa el resultado.

### La Regla de Oro

$$Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$$

**En español**:
- Si la recompensa fue mejor de lo esperado → aumenta Q(s,a)
- Si fue peor → disminuye Q(s,a)

## 2 - Crear Q-Table

La Q-table guarda el valor de cada acción en cada estado.

**Forma**: `(n_estados, n_acciones)`

Para FrozenLake: `(16, 4)`

<a name='ex-1'></a>
### Ejercicio 1 - crear_q_table

Crea una Q-table inicializada en ceros.

````python
# GRADED FUNCTION: crear_q_table

def crear_q_table(n_estados, n_acciones):
    """
    Crea Q-table inicializada en ceros.
    
    Argumentos:
    n_estados -- número de estados
    n_acciones -- número de acciones
    
    Retorna:
    Q -- array de forma (n_estados, n_acciones) con ceros
    """
    # (approx. 1 línea)
    # Usa np.zeros() con forma (n_estados, n_acciones)
    
    # TU CÓDIGO EMPIEZA AQUÍ
    Q = np.zeros((n_estados, n_acciones))
    # TU CÓDIGO TERMINA AQUÍ
    
    return Q
````

## 3 - Epsilon-Greedy: Explorar vs Explotar

El agente necesita **balancear**:
- **Explotar**: Usar lo que ya sabe (elegir mejor acción)
- **Explorar**: Probar cosas nuevas (elegir acción aleatoria)

### Estrategia Epsilon-Greedy

```python
if random < epsilon:
    accion = aleatoria()  # Explorar
else:
    accion = mejor()      # Explotar
```

**Ejemplo**: Si `epsilon = 0.1`
- 10% del tiempo → acción aleatoria
- 90% del tiempo → mejor acción

<a name='ex-2'></a>
### Ejercicio 2 - elegir_accion

Implementa epsilon-greedy para elegir acciones.

````python
# GRADED FUNCTION: elegir_accion

def elegir_accion(Q, estado, epsilon):
    """
    Elige acción usando epsilon-greedy.
    
    Argumentos:
    Q -- Q-table
    estado -- estado actual
    epsilon -- probabilidad de exploración (0 a 1)
    
    Retorna:
    accion -- acción elegida
    """
    # (approx. 5 líneas)
    # 1. Generar número aleatorio entre 0 y 1
    # 2. Si random < epsilon:
    #       accion = aleatoria (usa np.random.randint)
    #    Sino:
    #       accion = mejor (usa np.argmax en Q[estado])
    
    # TU CÓDIGO EMPIEZA AQUÍ
    if np.random.random() < epsilon:
        accion = np.random.randint(Q.shape[1])
    else:
        accion = np.argmax(Q[estado])
    # TU CÓDIGO TERMINA AQUÍ
    
    return accion
````

````python
# Prueba tu implementación
Q_test = np.array([[0.1, 0.5, 0.2, 0.3],  # Estado 0: mejor acción = 1
                   [0.8, 0.2, 0.3, 0.1]]) # Estado 1: mejor acción = 0

print("Probando con epsilon = 0.0 (siempre explotar):")
for _ in range(5):
    accion = elegir_accion(Q_test, estado=0, epsilon=0.0)
    print(f"  Acción elegida: {accion} (debería ser siempre 1)")

print("\nProbando con epsilon = 1.0 (siempre explorar):")
acciones = [elegir_accion(Q_test, estado=0, epsilon=1.0) for _ in range(20)]
print(f"  Acciones variadas: {len(set(acciones)) > 1}")

# Ejecuta el test
elegir_accion_test(elegir_accion)
````

## 4 - Actualizar Q-Value

Esta es la **regla de oro** de Q-Learning.

### La Fórmula

$$Q(s,a) \leftarrow Q(s,a) + \alpha \left[r + \gamma \max_{a'} Q(s',a') - Q(s,a)\right]$$

### Paso a Paso

1. **Valor actual**: `q_actual = Q[estado, accion]`
2. **Mejor valor futuro**: `max_q_siguiente = max(Q[siguiente_estado])`
3. **Target**: `target = recompensa + gamma * max_q_siguiente`
4. **Error**: `error = target - q_actual`
5. **Actualizar**: `Q[estado, accion] += alpha * error`

**Caso especial**: Si el episodio terminó (`done=True`), no hay futuro:
- `target = recompensa` (sin parte de `gamma * max_q_siguiente`)

<a name='ex-3'></a>
### Ejercicio 3 - actualizar_q

Implementa la actualización de Q-Learning.

````python
# GRADED FUNCTION: actualizar_q

def actualizar_q(Q, estado, accion, recompensa, siguiente_estado, done, alpha=0.1, gamma=0.99):
    """
    Actualiza Q-value usando la regla de Q-Learning.
    
    Argumentos:
    Q -- Q-table
    estado -- estado actual
    accion -- acción tomada
    recompensa -- recompensa recibida
    siguiente_estado -- siguiente estado
    done -- True si episodio terminó
    alpha -- tasa de aprendizaje (default: 0.1)
    gamma -- factor de descuento (default: 0.99)
    
    Retorna:
    Q -- Q-table actualizada
    """
    # (approx. 6 líneas)
    # 1. Obtener q_actual = Q[estado, accion]
    # 2. Si done:
    #       target = recompensa
    #    Sino:
    #       target = recompensa + gamma * max(Q[siguiente_estado])
    # 3. Actualizar: Q[estado, accion] += alpha * (target - q_actual)
    
    # TU CÓDIGO EMPIEZA AQUÍ
    q_actual = Q[estado, accion]
    if done:
        target = recompensa
    else:
        target = recompensa + gamma * np.max(Q[siguiente_estado])
    Q[estado, accion] += alpha * (target - q_actual)
    # TU CÓDIGO TERMINA AQUÍ
    
    return Q
````

````python
# Prueba tu implementación
Q_test = np.zeros((3, 2))
Q_test[1] = [0.5, 0.3]  # Siguiente estado tiene valores

print("Caso 1: Episodio NO terminado")
print(f"Q[0,0] antes: {Q_test[0,0]}")
Q_test = actualizar_q(Q_test, estado=0, accion=0, recompensa=1.0, 
                      siguiente_estado=1, done=False, alpha=0.1, gamma=0.9)
print(f"Q[0,0] después: {Q_test[0,0]:.4f}")
print(f"Esperado: ~0.145 (porque target = 1.0 + 0.9*0.5 = 1.45)")

print("\nCaso 2: Episodio terminado")
Q_test2 = np.zeros((3, 2))
Q_test2 = actualizar_q(Q_test2, estado=0, accion=0, recompensa=1.0,
                       siguiente_estado=1, done=True, alpha=0.1, gamma=0.9)
print(f"Q[0,0] después: {Q_test2[0,0]:.4f}")
print(f"Esperado: 0.1 (porque target = 1.0, sin futuro)")

# Ejecuta el test
actualizar_q_test(actualizar_q)
````

## 5 - Entrenar Agente

¡Ahora juntamos todo! El loop de entrenamiento:

```
Para cada episodio:
    1. Resetear ambiente
    2. Mientras no termine:
        a. Elegir acción (epsilon-greedy)
        b. Tomar acción en ambiente
        c. Actualizar Q-table
        d. Ir al siguiente estado
```

<a name='ex-4'></a>
### Ejercicio 4 - entrenar_q_learning

Implementa el loop de entrenamiento completo.

````python
# GRADED FUNCTION: entrenar_q_learning

def entrenar_q_learning(env, n_episodios=1000, alpha=0.1, gamma=0.99, epsilon=1.0):
    """
    Entrena un agente Q-Learning.
    
    Argumentos:
    env -- ambiente de Gymnasium
    n_episodios -- número de episodios de entrenamiento
    alpha -- tasa de aprendizaje
    gamma -- factor de descuento
    epsilon -- probabilidad de exploración
    
    Retorna:
    Q -- Q-table entrenada
    recompensas -- lista de recompensas totales por episodio
    """
    # Inicializar Q-table
    Q = crear_q_table(env.observation_space.n, env.action_space.n)
    recompensas = []
    
    # Loop de entrenamiento (approx. 15 líneas)
    # Para cada episodio:
    #   1. estado = env.reset()[0]
    #   2. done = False
    #   3. recompensa_total = 0
    #   4. Mientras not done:
    #      a. accion = elegir_accion(Q, estado, epsilon)
    #      b. siguiente_estado, recompensa, terminated, truncated, _ = env.step(accion)
    #      c. done = terminated or truncated
    #      d. Q = actualizar_q(Q, estado, accion, recompensa, siguiente_estado, done, alpha, gamma)
    #      e. recompensa_total += recompensa
    #      f. estado = siguiente_estado
    #   5. recompensas.append(recompensa_total)
    
    # TU CÓDIGO EMPIEZA AQUÍ
    for episodio in range(n_episodios):
        estado = env.reset()[0]
        done = False
        recompensa_total = 0
        
        # Epsilon decay para mejor aprendizaje
        epsilon_actual = max(0.01, epsilon - (epsilon - 0.01) * episodio / n_episodios)
        
        while not done:
            accion = elegir_accion(Q, estado, epsilon_actual)
            siguiente_estado, recompensa, terminated, truncated, _ = env.step(accion)
            done = terminated or truncated
            Q = actualizar_q(Q, estado, accion, recompensa, siguiente_estado, done, alpha, gamma)
            recompensa_total += recompensa
            estado = siguiente_estado
        
        recompensas.append(recompensa_total)
    # TU CÓDIGO TERMINA AQUÍ
    
    return Q, recompensas
````

````python
# Prueba tu implementación
print("Entrenando agente Q-Learning...")
print("Esto puede tomar 10-20 segundos...\n")

Q_entrenada, recompensas = entrenar_q_learning(
    env, 
    n_episodios=2000, 
    alpha=0.1, 
    gamma=0.99, 
    epsilon=1.0
)

print("✅ Entrenamiento completado!")
print(f"\nÚltimos 100 episodios:")
print(f"  Recompensa promedio: {np.mean(recompensas[-100:]):.2f}")
print(f"  Tasa de éxito: {np.mean(recompensas[-100:]) * 100:.0f}%")

# Ejecuta el test
entrenar_q_learning_test(entrenar_q_learning)
````

## 6 - Visualizar Resultados

Veamos cómo mejoró el agente durante el entrenamiento.

````python
# Graficar progreso de aprendizaje
plt.figure(figsize=(12, 5))

# Gráfico 1: Recompensas por episodio
plt.subplot(1, 2, 1)
plt.plot(recompensas, alpha=0.3, label='Recompensa por episodio')

# Promedio móvil
ventana = 50
if len(recompensas) >= ventana:
    promedio_movil = np.convolve(recompensas, np.ones(ventana)/ventana, mode='valid')
    plt.plot(range(ventana-1, len(recompensas)), promedio_movil, 
             label=f'Promedio móvil ({ventana} eps)', linewidth=2, color='red')

plt.xlabel('Episodio')
plt.ylabel('Recompensa')
plt.title('Progreso de Aprendizaje')
plt.legend()
plt.grid(True, alpha=0.3)

# Gráfico 2: Tasa de éxito por bloques
plt.subplot(1, 2, 2)
bloques = 20
tamano_bloque = len(recompensas) // bloques
tasas_exito = []
for i in range(bloques):
    inicio = i * tamano_bloque
    fin = (i + 1) * tamano_bloque
    tasa = np.mean(recompensas[inicio:fin]) * 100
    tasas_exito.append(tasa)

plt.plot(tasas_exito, marker='o', linewidth=2)
plt.xlabel(f'Bloque (cada {tamano_bloque} episodios)')
plt.ylabel('Tasa de Éxito (%)')
plt.title('Tasa de Éxito por Bloques')
plt.grid(True, alpha=0.3)
plt.ylim([0, 105])

plt.tight_layout()
plt.show()

print(f"\n📊 Análisis:")
print(f"  Primeros 100 episodios: {np.mean(recompensas[:100]) * 100:.0f}% éxito")
print(f"  Últimos 100 episodios: {np.mean(recompensas[-100:]) * 100:.0f}% éxito")
print(f"  Mejora: +{(np.mean(recompensas[-100:]) - np.mean(recompensas[:100])) * 100:.0f}%")
````

### Probar el Agente Entrenado

````python
# Evaluar agente (sin exploración)
print("Evaluando agente entrenado (100 episodios)...\n")

exitos = 0
for _ in range(100):
    estado = env.reset()[0]
    done = False
    
    while not done:
        # Siempre elegir mejor acción (epsilon=0)
        accion = np.argmax(Q_entrenada[estado])
        siguiente_estado, recompensa, terminated, truncated, _ = env.step(accion)
        done = terminated or truncated
        estado = siguiente_estado
        
        if recompensa > 0:
            exitos += 1
            break

print(f"🎯 Tasa de éxito: {exitos}% (sin exploración)")
print(f"\n¡El agente aprendió a resolver FrozenLake!")
````
