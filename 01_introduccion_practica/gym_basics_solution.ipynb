# Gymnasium Basics - Fundamentos de Ambientes RL

¡Bienvenido! En este notebook aprenderás a trabajar con **Gymnasium** (antes conocido como OpenAI Gym), la biblioteca estándar para ambientes de Reinforcement Learning.

## ¿Qué es Gymnasium?

Gymnasium proporciona **ambientes** donde tu agente puede:
- 🎮 Observar el **estado** actual
- 🎯 Tomar **acciones**
- 🎁 Recibir **recompensas**
- 🔄 Aprender de la experiencia

Es como un videojuego donde tu código es el jugador.

## El Ciclo de RL

```
┌─────────────┐
│  Ambiente   │
└──────┬──────┘
       │ estado
       ↓
┌─────────────┐
│   Agente    │
└──────┬──────┘
       │ acción
       ↓
┌─────────────┐
│  Ambiente   │
└──────┬──────┘
       │ recompensa + nuevo estado
       ↓
    (repetir)
```

## 2 - Crear y Explorar Ambiente

Vamos a usar **FrozenLake**, un ambiente simple perfecto para aprender.

### FrozenLake: El Juego

Imagina un lago congelado con un patrón 4×4:

```
S F F F
F H F H
F F F H
H F F G
```

- **S** = Start (inicio)
- **F** = Frozen (congelado, seguro)
- **H** = Hole (agujero, ¡pierdes!)
- **G** = Goal (meta, ¡ganas!)

**Objetivo**: Ir de S a G sin caer en H.

### Crear el Ambiente

```python
env = gym.make('FrozenLake-v1', is_slippery=False)
```

- `'FrozenLake-v1'`: nombre del ambiente
- `is_slippery=False`: movimientos deterministas (más fácil para aprender)

### Espacios

Cada ambiente tiene:
- **Espacio de observación** (`observation_space`): posibles estados
- **Espacio de acciones** (`action_space`): posibles acciones

<a name='ex-1'></a>
### Ejercicio 1 - explorar_ambiente

Crea un ambiente y obtén el número de estados y acciones.

````python
# GRADED FUNCTION: explorar_ambiente

def explorar_ambiente(nombre_ambiente):
    """
    Crea un ambiente y retorna el número de estados y acciones.
    
    Argumentos:
    nombre_ambiente -- string con el nombre del ambiente (ej: 'FrozenLake-v1')
    
    Retorna:
    n_estados -- número de estados posibles
    n_acciones -- número de acciones posibles
    """
    # (approx. 3 líneas)
    # 1. Crear ambiente con gym.make(nombre_ambiente, is_slippery=False)
    # 2. Obtener n_estados de env.observation_space.n
    # 3. Obtener n_acciones de env.action_space.n
    
    # TU CÓDIGO EMPIEZA AQUÍ
    env = gym.make(nombre_ambiente, is_slippery=False)
    n_estados = env.observation_space.n
    n_acciones = env.action_space.n
    # TU CÓDIGO TERMINA AQUÍ
    
    return n_estados, n_acciones
````

````python
# Prueba tu implementación
n_estados, n_acciones = explorar_ambiente('FrozenLake-v1')

print(f"FrozenLake tiene:")
print(f"  - {n_estados} estados (posiciones en el grid 4×4)")
print(f"  - {n_acciones} acciones (izquierda, abajo, derecha, arriba)")

# Ejecuta el test
explorar_ambiente_test(explorar_ambiente)
````

## 3 - Reset: Empezar un Episodio

Cada vez que queremos empezar un nuevo episodio (intento), llamamos a `env.reset()`.

**Sintaxis**:
```python
estado_inicial, info = env.reset()
```

- `estado_inicial`: el estado donde empieza el agente (usualmente 0 en FrozenLake)
- `info`: información adicional (puedes ignorarla por ahora)

<a name='ex-2'></a>
### Ejercicio 2 - resetear_ambiente

Resetea un ambiente y retorna el estado inicial.

````python
# GRADED FUNCTION: resetear_ambiente

def resetear_ambiente(env):
    """
    Resetea el ambiente y retorna el estado inicial.
    
    Argumentos:
    env -- ambiente de Gymnasium
    
    Retorna:
    estado_inicial -- estado donde empieza el episodio
    """
    # (approx. 1 línea)
    # Llama env.reset() y obtén el primer valor (estado_inicial)
    # Recuerda que reset() retorna (estado, info)
    
    # TU CÓDIGO EMPIEZA AQUÍ
    estado_inicial, _ = env.reset()
    # TU CÓDIGO TERMINA AQUÍ
    
    return estado_inicial
````

````python
# Prueba tu implementación
env = gym.make('FrozenLake-v1', is_slippery=False)
estado = resetear_ambiente(env)

print(f"Estado inicial: {estado}")
print(f"Esperado: 0 (esquina superior izquierda)")

# Ejecuta el test
resetear_ambiente_test(resetear_ambiente)
````

**Output esperado**:
```
Estado inicial: 0
Esperado: 0 (esquina superior izquierda)
✅ Test pasado!
```

## 4 - Step: Tomar una Acción

Para interactuar con el ambiente, usamos `env.step(accion)`.

**Sintaxis**:
```python
siguiente_estado, recompensa, terminated, truncated, info = env.step(accion)
```

**Retorna 5 valores**:
1. `siguiente_estado`: nuevo estado después de la acción
2. `recompensa`: recompensa recibida (1 si llegas a la meta, 0 si no)
3. `terminated`: True si el episodio terminó (llegaste a meta o agujero)
4. `truncated`: True si se alcanzó límite de pasos
5. `info`: información adicional

**Ejemplo**:
```python
# Estamos en estado 0, tomamos acción 2 (derecha)
siguiente_estado, recompensa, terminated, truncated, info = env.step(2)
# Resultado: siguiente_estado = 1, recompensa = 0, terminated = False
```

<a name='ex-3'></a>
### Ejercicio 3 - tomar_accion

Toma una acción en el ambiente y retorna el resultado.

````python
# GRADED FUNCTION: tomar_accion

def tomar_accion(env, accion):
    """
    Toma una acción en el ambiente.
    
    Argumentos:
    env -- ambiente de Gymnasium
    accion -- acción a tomar (0, 1, 2, o 3)
    
    Retorna:
    siguiente_estado -- nuevo estado
    recompensa -- recompensa recibida
    done -- True si el episodio terminó
    """
    # (approx. 2 líneas)
    # 1. Llama env.step(accion) y obtén los 5 valores
    # 2. Combina terminated y truncated en done = terminated or truncated
    
    # TU CÓDIGO EMPIEZA AQUÍ
    siguiente_estado, recompensa, terminated, truncated, info = env.step(accion)
    done = terminated or truncated
    # TU CÓDIGO TERMINA AQUÍ
    
    return siguiente_estado, recompensa, done
````

````python
# Prueba tu implementación
env = gym.make('FrozenLake-v1', is_slippery=False)
estado = env.reset()[0]

print(f"Estado inicial: {estado}")
print("\nTomando acción 2 (derecha)...")

siguiente_estado, recompensa, done = tomar_accion(env, 2)

print(f"  Siguiente estado: {siguiente_estado}")
print(f"  Recompensa: {recompensa}")
print(f"  Episodio terminado: {done}")

# Ejecuta el test
tomar_accion_test(tomar_accion)
````

**Output esperado**:
```
Estado inicial: 0

Tomando acción 2 (derecha)...
  Siguiente estado: 1
  Recompensa: 0.0
  Episodio terminado: False
✅ Test pasado!
```

## 5 - Episodio Completo

Un **episodio** es una secuencia completa desde el inicio hasta que termina (llegamos a la meta o caemos en un agujero).

**Estructura básica**:
```python
estado = env.reset()[0]
done = False

while not done:
    accion = ... # elegir acción
    siguiente_estado, recompensa, done = tomar_accion(env, accion)
    estado = siguiente_estado
```

<a name='ex-4'></a>
### Ejercicio 4 - ejecutar_episodio_random

Ejecuta un episodio completo con acciones aleatorias.

````python
# GRADED FUNCTION: ejecutar_episodio_random

def ejecutar_episodio_random(env, max_pasos=100):
    """
    Ejecuta un episodio completo con acciones aleatorias.
    
    Argumentos:
    env -- ambiente de Gymnasium
    max_pasos -- máximo número de pasos por episodio
    
    Retorna:
    recompensas -- lista de recompensas recibidas en cada paso
    """
    # (approx. 8 líneas)
    # 1. Resetear ambiente
    # 2. Inicializar lista de recompensas vacía
    # 3. Inicializar done = False
    # 4. Loop while not done:
    #    a. Elegir acción aleatoria: env.action_space.sample()
    #    b. Tomar acción (usa tu función tomar_accion)
    #    c. Guardar recompensa en la lista
    #    d. Actualizar estado
    # 5. Retornar lista de recompensas
    
    # TU CÓDIGO EMPIEZA AQUÍ
    estado = env.reset()[0]
    recompensas = []
    done = False
    pasos = 0
    while not done and pasos < max_pasos:
        accion = env.action_space.sample()
        siguiente_estado, recompensa, done = tomar_accion(env, accion)
        recompensas.append(recompensa)
        estado = siguiente_estado
        pasos += 1
    # TU CÓDIGO TERMINA AQUÍ
    
    return recompensas
````

````python
# Prueba tu implementación
env = gym.make('FrozenLake-v1', is_slippery=False)

print("Ejecutando 3 episodios con acciones aleatorias...\n")

for i in range(3):
    recompensas = ejecutar_episodio_random(env)
    total = sum(recompensas)
    print(f"Episodio {i+1}: {len(recompensas)} pasos, recompensa total = {total}")

# Ejecuta el test
ejecutar_episodio_random_test(ejecutar_episodio_random)
````

**Output esperado** (los números variarán porque son aleatorios):
```
Ejecutando 3 episodios con acciones aleatorias...

Episodio 1: 15 pasos, recompensa total = 0.0
Episodio 2: 8 pasos, recompensa total = 0.0
Episodio 3: 23 pasos, recompensa total = 1.0
✅ Test pasado!
```

## 6 - Evaluar Performance

Para saber qué tan bien lo estamos haciendo, calculamos la **recompensa total** de un episodio.

<a name='ex-5'></a>
### Ejercicio 5 - calcular_recompensa_total

Calcula la recompensa total de una lista de recompensas.

````python
# GRADED FUNCTION: calcular_recompensa_total

def calcular_recompensa_total(recompensas):
    """
    Calcula la recompensa total de un episodio.
    
    Argumentos:
    recompensas -- lista de recompensas
    
    Retorna:
    total -- suma de todas las recompensas
    """
    # (approx. 1 línea)
    # Usa sum() para sumar todas las recompensas
    
    # TU CÓDIGO EMPIEZA AQUÍ
    total = sum(recompensas)
    # TU CÓDIGO TERMINA AQUÍ
    
    return total
````

````python
# Prueba tu implementación
env = gym.make('FrozenLake-v1', is_slippery=False)

print("Ejecutando 10 episodios y calculando recompensa total...\n")

totales = []
for i in range(10):
    recompensas = ejecutar_episodio_random(env)
    total = calcular_recompensa_total(recompensas)
    totales.append(total)
    print(f"Episodio {i+1}: {total}")

exitos = sum(1 for t in totales if t > 0)
print(f"\nÉxitos: {exitos}/10 ({exitos*10}%)")

# Ejecuta el test
calcular_recompensa_total_test(calcular_recompensa_total)
````

**Output esperado** (los números variarán):
```
Ejecutando 10 episodios y calculando recompensa total...

Episodio 1: 0.0
Episodio 2: 1.0
Episodio 3: 0.0
...

Éxitos: 2/10 (20%)
✅ Test pasado!
```

### 💡 Observación Importante

Con acciones **aleatorias**, la tasa de éxito es muy baja (~10-20%).
